```python
# Load each game's Reviews sheet and show how many rows it contains.
import pandas as pd

destiny = pd.read_excel(
    "7_Destiny_2_1085660xlsx.xlsx",
    sheet_name="Reviews"
)

warframe = pd.read_excel(
    "16_Warframe_230410.xlsx",
    sheet_name="Reviews"
)

print("Destiny reviews:", len(destiny))
print("Warframe reviews:", len(warframe))
```

```python
# Preview Destiny review text and its recommendation label.
destiny[["review", "voted_up"]].head()
```

```python
# Select one Destiny review to demonstrate the cleaning steps.
import string

review = destiny["review"][2]

print(review)
```

```python
# Split the example into words, trim punctuation, and lowercase the text.
words = review.split()

clean_words = []

for word in words:
    word_clean = word.strip(string.punctuation)
    word_clean = word_clean.lower()
    clean_words.append(word_clean)

print(clean_words)
```

```python
# Count the words in the example and show the 10 most common.
from collections import Counter

word_counts = Counter(clean_words)

print(word_counts.most_common(10))
```

```python
# Look up how often 'game' appears in the example review.
print("Game mentions:", word_counts["game"])
```

```python
# Remove common words so the remaining word counts are easier to interpret.
stopwords = [
    "a", "an", "the", "it", "of", "and",
    "is", "was", "to", "in", "for", "on"
]

filtered_words = []

for word in clean_words:
    if word not in stopwords:
        filtered_words.append(word)

filtered_counts = Counter(filtered_words)

print(filtered_counts.most_common(10))
```

```python
# Define a reusable function to clean each review and remove stopwords.
def clean_review(review):
    filtered_words = []

    review = review.replace("·", " ")

    for word in review.split():
        word_clean = word.strip(string.punctuation).lower()

        if word_clean != "" and word_clean not in stopwords:
            filtered_words.append(word_clean)

    return filtered_words
```

```python
# Check how the cleaning function handles a short test sentence.
test_words = clean_review("The gameplay is fun, but NOT perfect!")

print(test_words)
```

```python
# Clean all Destiny reviews with text and show the 20 most common words.
destiny_reviews = destiny["review"].dropna().tolist()

destiny_words = []

for review in destiny_reviews:
    cleaned = clean_review(review)
    destiny_words.extend(cleaned)

destiny_counts = Counter(destiny_words)

print("Reviews analysed:", len(destiny_reviews))
print(destiny_counts.most_common(20))
```

```python
# Split Destiny into recommended and not-recommended reviews with text.
destiny_positive = destiny[destiny["voted_up"] == True]
destiny_negative = destiny[destiny["voted_up"] == False]

positive_reviews = destiny_positive["review"].dropna().tolist()
negative_reviews = destiny_negative["review"].dropna().tolist()

print("Recommended reviews with text:", len(positive_reviews))
print("Not-recommended reviews with text:", len(negative_reviews))
```

```python
# Compare the most common words in the two Destiny review groups.
positive_words = []
negative_words = []

for review in positive_reviews:
    positive_words.extend(clean_review(review))

for review in negative_reviews:
    negative_words.extend(clean_review(review))

positive_counts = Counter(positive_words)
negative_counts = Counter(negative_words)

print("RECOMMENDED:")
print(positive_counts.most_common(15))

print("NOT RECOMMENDED:")
print(negative_counts.most_common(15))
```

```python
# Compare keyword counts per 1,000 cleaned words, allowing for unequal text totals.
keywords = ["gameplay", "fun", "bugs", "money", "content", "bungie"]

for word in keywords:
    positive_rate = positive_counts[word] / len(positive_words) * 1000
    negative_rate = negative_counts[word] / len(negative_words) * 1000

    print(word)
    print("  Recommended:", round(positive_rate, 2))
    print("  Not recommended:", round(negative_rate, 2))
```

```python
# Read the first three negative Destiny reviews mentioning 'money' in context.
shown = 0

for review in negative_reviews:
    words = clean_review(review)

    if "money" in words:
        print(review)
        print("---")

        shown = shown + 1

        if shown == 3:
            break
```

```python
# Read the first three positive Destiny reviews mentioning 'money' in context.
shown = 0

for review in positive_reviews:
    words = clean_review(review)

    if "money" in words:
        print(review)
        print("---")

        shown = shown + 1

        if shown == 3:
            break
```

```python
# Count reviews mentioning 'money' once each and report each Destiny group's percentage.
positive_money = 0
negative_money = 0

for review in positive_reviews:
    if "money" in clean_review(review):
        positive_money = positive_money + 1

for review in negative_reviews:
    if "money" in clean_review(review):
        negative_money = negative_money + 1

positive_percent = positive_money / len(positive_reviews) * 100
negative_percent = negative_money / len(negative_reviews) * 100

print("Recommended:", positive_money, "reviews")
print("Percentage:", round(positive_percent, 2))

print("Not recommended:", negative_money, "reviews")
print("Percentage:", round(negative_percent, 2))
```

```python
# Split Warframe into recommended and not-recommended reviews with text.
warframe_positive = warframe[warframe["voted_up"] == True]
warframe_negative = warframe[warframe["voted_up"] == False]

warframe_positive_reviews = warframe_positive["review"].dropna().tolist()
warframe_negative_reviews = warframe_negative["review"].dropna().tolist()

print("Warframe recommended:", len(warframe_positive_reviews))
print("Warframe not recommended:", len(warframe_negative_reviews))
```

```python
# Repeat the 'money' review counts and percentages for Warframe.
warframe_positive_money = 0
warframe_negative_money = 0

for review in warframe_positive_reviews:
    if "money" in clean_review(review):
        warframe_positive_money += 1

for review in warframe_negative_reviews:
    if "money" in clean_review(review):
        warframe_negative_money += 1

warframe_positive_percent = (
    warframe_positive_money / len(warframe_positive_reviews) * 100
)

warframe_negative_percent = (
    warframe_negative_money / len(warframe_negative_reviews) * 100
)

print("Recommended:", warframe_positive_money, "reviews")
print("Percentage:", round(warframe_positive_percent, 2))

print("Not recommended:", warframe_negative_money, "reviews")
print("Percentage:", round(warframe_negative_percent, 2))
```

```python
# Print every negative Warframe review mentioning 'money' to check its context.
for review in warframe_negative_reviews:
    if "money" in clean_review(review):
        print(review)
        print("---")
```

```python
# Print the first three positive Warframe reviews mentioning 'money'.
shown = 0

for review in warframe_positive_reviews:
    if "money" in clean_review(review):
        print(review)
        print("---")
        shown += 1

        if shown == 3:
            break
```

Destiny 2: “money” appears in 3.07% of recommended reviews and 17.86% of not-recommended reviews.
Warframe: the corresponding figures are 2.24% and 1.67%.
Warframe’s only not-recommended review mentioning “money” concerns a ban and support, rather than pricing.
The recommended examples contain mixed attitudes toward spending.
This search only matches “money”. It misses related terms such as “price,” “paid,” and “expensive.”
The first three matching reviews are illustrative examples, not a random sample.

```python
# Define broader spending keywords and count each matching review once.
spending_words = [
    "money", "price", "prices", "pricing",
    "pay", "paid", "paying", "spend", "spent",
    "spending", "expensive", "cheap",
    "purchase", "purchases", "monetization",
    "microtransactions", "paywall", "paywalls"
]

def count_spending_reviews(reviews):
    count = 0

    for review in reviews:
        words = clean_review(review)

        for keyword in spending_words:
            if keyword in words:
                count += 1
                break

    return count
```

```python
# Compare spending-keyword review counts and percentages across all four groups.
review_groups = {
    "Destiny recommended": positive_reviews,
    "Destiny not recommended": negative_reviews,
    "Warframe recommended": warframe_positive_reviews,
    "Warframe not recommended": warframe_negative_reviews
}

for name, reviews in review_groups.items():
    count = count_spending_reviews(reviews)
    percentage = count / len(reviews) * 100

    print(name)
    print("Matching reviews:", count)
    print("Percentage:", round(percentage, 2))
    print("---")
```

```python
# Read negative Warframe keyword matches to check what the spending words refer to.
for review in warframe_negative_reviews:
    words = clean_review(review)

    for keyword in spending_words:
        if keyword in words:
            print(review)
            print("---")
            break
```

Manual inspection found that neither of the two matching not-recommended Warframe reviews clearly criticised pricing. One used “spend” to discuss time, while the other mentioned money invested when discussing a ban. This demonstrates that keyword matches alone cannot reliably identify monetization complaints.

```python
# Sample five matching negative Destiny reviews; seed 42 makes the selection repeatable.
import random

destiny_negative_matches = []

for review in negative_reviews:
    words = clean_review(review)

    for keyword in spending_words:
        if keyword in words:
            destiny_negative_matches.append(review)
            break

sample_reviews = random.Random(42).sample(
    destiny_negative_matches, 5
)

for number, review in enumerate(sample_reviews, start=1):
    print("REVIEW", number)
    print(review)
    print("---")
```

```python
# Record manual labels for the five Destiny reviews; True means a monetization complaint.
sample_check = pd.DataFrame({
    "review_text": sample_reviews,
    "monetization_complaint": [True, True, True, True, True],
    "reason": [
        "Repeated purchases, paywalls, and removal of paid content",
        "Explicit criticism of monetization",
        "Opposes further DLC spending because of dissatisfaction",
        "Criticises paid cosmetics and emphasis on the store",
        "Criticises repeated charges and removal of purchased content"
    ]
})

sample_check
```

```python
# Save and download the manually checked Destiny sample as a CSV.
sample_check.to_csv(
    "destiny_negative_sample_check.csv",
    index=False
)

files.download("destiny_negative_sample_check.csv")
```

## Warframe: negative review check and CSV output

This repeats the Destiny check using the same `clean_review` function and
`spending_words` list. The current data has only two matching negative
Warframe reviews, so both are included rather than sampling five.

The methods follow the course notebooks on
[loops and lists](https://github.com/mervetekgurler/cts2000_fall2026/blob/main/units/week_2/2-2_iteration.ipynb),
[pandas tables](https://github.com/mervetekgurler/cts2000_fall2026/blob/main/units/week_6/6-1_pandas.ipynb), and
[text cleaning and CSV export](https://github.com/mervetekgurler/cts2000_fall2026/blob/main/units/week_6/6-2_corpus_level_text_analysis_montagu.ipynb).


```python
# Collect and print all matching negative Warframe reviews; two match in this dataset.
warframe_negative_matches = []

for review in warframe_negative_reviews:
    words = clean_review(review)

    for keyword in spending_words:
        if keyword in words:
            warframe_negative_matches.append(review)
            break

print("Warframe matching negative reviews:", len(warframe_negative_matches))

shown = 0

for review in warframe_negative_matches:
    shown += 1
    print("WARFRAME REVIEW", shown)
    print(review)
    print("---")
```

### Manual labels for the two reviews above

These labels are manual interpretations, just like the Destiny labels:

1. The New War review criticises quest design and uses "spend" about time
   spent waiting, rather than the cost of playing.
2. The ban review mentions money invested, but its complaint concerns the
   anti-cheat system and support rather than pricing.

Neither is coded as a clear monetization complaint. The labels and reasons
below match these two reviews in the order printed above; review them again
if the source data changes.


```python
# Record manual labels for the two Warframe reviews, in the order printed above.
warframe_sample_check = pd.DataFrame({
    "review_text": warframe_negative_matches,
    "monetization_complaint": [False, False],
    "reason": [
        "Criticises quest design; uses spend about time waiting, not money",
        "Criticises a ban and support; mentions money invested, not pricing"
    ]
})

warframe_sample_check
```

```python
# Save and download the manually checked Warframe reviews as a CSV.
warframe_sample_check.to_csv(
    "warframe_negative_sample_check.csv",
    index=False
)

files.download("warframe_negative_sample_check.csv")
```
